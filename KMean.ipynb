```python
from sklearn.cluster import KMeans
```

```python
# use elbow method for find k value 
wcss=[]
for i in range(1,11):
    model = KMeans(n_clusters=i,random_state=42)
    model.fit(df)
    wcss.append(model.inertia_)
wcss
```

```python
plt.plot(range(1,11),wcss)  # it give 3 cluster we have to take so analysis
```
